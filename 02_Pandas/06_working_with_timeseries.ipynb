#Working with Time Series

```python
import pandas as pd

data = pd.DataFrame({
    'Date': [
        '2025-04-28',
        '2025-04-29',
        '2025-04-30',
        '2025-05-01',
        '2025-05-02'
    ],
    'Lemonades': [10, 20, 15, 25, 30]
})

print(data)
```

#DateTime Index


A DateTime index means using a date column as the DataFrame's index. This makes it easier to select, filter, and summarize data by date.

```python
#This converts date strings into Pandas datetime values so Pandas can recognize and process them as dates.
data['Date'] = pd.to_datetime(data['Date'])

#Set the date as the index.
data.set_index('Date', inplace=True)
```

#Resampling

groups time-based data into a different time frequency, such as daily, weekly, monthly, or yearly, and lets you calculate summaries for each period.

```python
data.resample('W').sum()
```

```python
data.resample('M').sum()
```

#Shifting / Lagging

shift() moves values up or down by a specified number of rows. It is often used to compare the current value with a previous value.

```python
data['Previous Lemonades'] = data['Lemonades'].shift(1)
```

```python
data
```

#Rolling Calculations — rolling()

rolling() calculates a statistic over a moving window of rows, such as a rolling average over the last two observations.

```python
data['Rolling Average'] = (
    data['Lemonades'].rolling(window=2).mean()
)

#rolling(window=2) considers two consecutive values at a time.
#.mean() calculates their average.
#The window moves forward one row at a time.
```
