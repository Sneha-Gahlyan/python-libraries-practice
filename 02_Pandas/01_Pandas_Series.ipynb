#Methods

```python

sales = pd.Series([500, 200, 800, None, 400, 900])
print(sales)
```

###1. sales.head()
Returns the first 5 elements of a Series by default.

```python
sales.head()
```

```python
sales.head(6)
```

###sales.tail()
 Returns the last five elements of a Series by default.

```python
sales.tail(3)
```

###sort_values()
Sorts the values in a Series in ascending or descending order.

```python
sales.sort_values()
```

series.sort_values(
    ascending=True,
    na_position="last"
)

na_position="last": Places missing values at the end by default.
na_position="first" → NaN at the start.


```python
sales.sort_values(na_position = "first")
```

4. count()
Returns the number of non-missing values in a Series.

len(sales) counts all elements, including missing ones.




```python
print(sales.count())
```

isnull()
Checks each element to determine whether it is missing (NaN or another recognized missing value). Returns True for missing values and False otherwise.

```python
print(sales.isnull())
```

#Missing Values

```python

sales_with_missing = pd.Series([2, None, 5])

print(sales_with_missing)
```

```python
sales_with_missing.isnull()
```

fillna(value)


Replaces missing values with the value you specify.

series.fillna(value)

```python
print(sales_with_missing.fillna(0))
```

By default, fillna() returns a new Series and does not modify the original Series.

To update the original variable, assign the result back:

```python
sales_with_missing = sales_with_missing.fillna(0)
```
