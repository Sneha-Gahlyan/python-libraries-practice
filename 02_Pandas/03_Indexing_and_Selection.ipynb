#Indexing and Selection

###1. data.set_index('Name', inplace=True)

set_index() is used to make an existing column the index (row label) of a DataFrame instead of using the default numbers like 0, 1, 2

```python
import pandas as pd

s1 = pd.Series(data=[1,2,3,4,56])
s2 = pd.Series(data=[8,9,5,4,3])

data = pd.DataFrame({
    'C1': s1,
    'C2': s2
    })

data.set_index('C1' , inplace=True)
print(data)
```

###2. reset_index()


reset_index() is used to bring the existing index back to a normal column and create a new default index (0, 1, 2, ...).

```python
data.reset_index(inplace=True)
print(data)
```

###3. MultiIndex

A MultiIndex means a DataFrame has more than one index level. Instead of using only one column as the row label, we use two or more columns together.

```python
multi_data = pd.DataFrame({
    'Park': ['A', 'A', 'B', 'B'],
    'Name': ['Anna', 'Ben', 'Chris', 'Daisy'],
    'Lemonades': [2, 3, 4, 5]
}).set_index(['Park', 'Name'])

print(multi_data)
```

###4. Selecting with Conditions

 means filtering rows based on a rule.

```python
multi_data[multi_data['Lemonades'] >= 3]
```

###5. Query Method — query()


The query() method filters rows using a condition written as a string.

```python
multi_data.query('Lemonades >= 3')
```

###2. Creating a DataFrame from a Dictionary

###3. Creating a DataFrame from a List


 A list of lists can be converted into a DataFrame, where each inner list represents one row.

```python
import pandas as pd
listt = [
    [1, 2, 100],
    [2, 4, 100],
    [3, 8, 100]
]

pd.DataFrame(listt, columns=['x', 'y', 'z'])
```
