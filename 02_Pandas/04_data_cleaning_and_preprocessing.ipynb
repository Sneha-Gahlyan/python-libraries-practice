#Data Cleaning and Preprocessing

###1. Handling missing data


1. dropna() does not modify the original DataFrame by default. To save the change:

```python
import pandas as pd
```

```python
saless = pd.Series([2,4,6,None,3,None,3])

saless.dropna()
```

2. .fillna(0)

Replaces missing values with a specified value. Here, 0 is used.

```python
saless.fillna(0,inplace = True)
```

```python
saless
```

* Fill witth mean

```python
saless.mean()
```

```python
saless.fillna(saless.mean(),inplace = True)
```

```python
saless
```

#Removing Duplicates


drop_duplicates() removes duplicate rows from a DataFrame, keeping the first occurrence by default.


```python
data = pd.DataFrame({
    'C1': ['Anna', 'Ben', 'Anna', 'Raj'],
    'C2': [2, 3, 2, 5]
})

```

```python
data.drop_duplicates()
```

```python
data
```

###Replacing Duplicates

```python
data.replace(2,900)
```

```python
data
```

#Type Conversion

astype()
