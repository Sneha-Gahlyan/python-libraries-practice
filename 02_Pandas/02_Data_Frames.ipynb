#Pandas DataFrames

A DataFrame is a two-dimensional data structure in Pandas that stores data in rows and columns

```python
import pandas as pd
```

```python
# create data frame

# pd.DataFrame({'column_name': [values]})

data = pd.DataFrame({
    'Name': ['Anna', 'Ben'],
    'Lemonades': [2, 3]
})

print(data)
```

```python

s1 = pd.Series(data=[1, 2, 3, 4, 5])

print(s1)

print()

s2 = pd.Series(data=[10, 20, 30, 40, 50])

print(s2)
```

Creating a Series with custom index labels

```python
si = pd.Series(
    data=[100, 200, 300, 400, 500],
    index=list('abcde')
)

print(si)
```

Creating a DataFrame from two Series

```python
data = pd.DataFrame({
    'C1': s1,
    'C2': s2
})

print(data)

#'C1': s1 creates the first column named C1.
#'C2': s2 creates the second column named C2.
#Pandas aligns the Series using their index labels.
```

#Accessing data in Dataframe

loc[] → Access data using row and column labels.



iloc[] → Access data using integer positions.

```python
# data.loc[row_label, column_label]

print(data.loc[1, 'C1'])
```

```python
# data.iloc[row_position, column_position]

print(data.iloc[1, 0])
```

# Updating a value


- data[C1] selects the column whose name is the C1.
- data[C1] == 25 finds rows where that column's value is 25.
- data.loc[condition, 'C1'] = 19 updates the C1 value in those rows to 19.


```python
data
```

```python
data.loc[data['C1'] == 3] = 22

data
```

```python
data.loc[data['C1'] == 22,'C1'] = 23

data
```
