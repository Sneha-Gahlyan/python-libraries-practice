#Data Visualization

```python
import pandas as pd

data = pd.DataFrame({
    'Date': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Lemonades': [10, 20, 15, 25, 30]
})

```

```python

data.set_index('Date', inplace=True)
```

```python
data
```

#Plotting

```python
data['Lemonades'].plot()
```

###Different kinds of plots

1. line plot

```python
data['Lemonades'].plot(kind='line')
```

2. Bar plot

```python
data['Lemonades'].plot(kind='bar')
```

*fiigsize()

width , height

```python
data['Lemonades'].plot(kind='bar' , figsize=(2,3))
```

3. bruh plot

```python
data['Lemonades'].plot(kind='barh')
```

4. hist plot

Shows the distribution or frequency of numerical values.

A histogram groups values into intervals called bins.

```python
data['Lemonades'].plot(kind='hist')
```

5. Box plot

```python
data['Lemonades'].plot(kind='box')
```

6. Scatter plot

```python
sales = pd.DataFrame({
    'Advertising': [100, 200, 300, 400, 500],
    'Lemonades': [10, 15, 20, 28, 35]
})

```

```python

sales.plot(kind='scatter', x='Advertising', y='Lemonades')
```

7. Area plot

```python
data['Lemonades'].plot(kind='area')
```

8. Pie chart

```python
data['Lemonades'].plot(kind='pie')
```
