```python
import pandas as pd
```

```python
import pandas as pd

data = pd.DataFrame({
    'Name': ['Anna', 'Ben', 'Anna', 'Ben', 'Chris'],
    'Lemonades': [2, 3, 4, 5, 6]
})

print(data)
```

#1. groupby()

groups rows that have the same value in a specified column.

```python
grouped = data.groupby('Name')
```

```python
data = grouped
```

```python
print(data)
```

```python
grouped.sum()
```

#Aggregation

```python
#1
grouped['Lemonades'].mean()
```

```python
#2
grouped['Lemonades'].min()
```

```python
#It returns the largest value in each group.
grouped['Lemonades'].max()
```

#pivot_table()

 A pivot table summarizes data by grouping it using selected rows and columns and applying an aggregation function.

```python
data.pivot_table(values='Lemonades', index='Name')
```

```python
data.pivot_table(
    values='Lemonades',
    index='Name',
    aggfunc=['sum','mean']
)
```

#pd.crosstab()

 pd.crosstab() creates a frequency table showing how often combinations of values occur.

```python
pd.crosstab(data['Name'], data['Lemonades'])
```

```python
import pandas as pd

data2 = pd.DataFrame({
    'Name': ['Anna', 'Ben', 'Chris', 'Ben', 'Chris'],
    'Lemonades': [12, 13, 4, 25, 96]
})
```

```python
data2
```

***Concatenation***

Concatenation combines DataFrames by stacking them vertically (rows) or horizontally (columns).

```python
result = pd.concat([data,data2])
```

```python
result
```

```python
#to create a fressh index:

result2 = pd.concat([data,data2], ignore_index=True)
```

```python
result2
```

```python
#horizontally
pd.concat([data,data2], axis=1)
```

***merge()***

it uses a inner join

Merging combines DataFrames by matching values in a common column or columns, similar to SQL joins.

```python

pd.merge(data, data2, on='Name')
```

```python
#change join type
pd.merge(data, data2, on='Name', how='inner')
pd.merge(data, data2, on='Name', how='left')
pd.merge(data, data2, on='Name', how='right')
pd.merge(data, data2, on='Name', how='outer')
```

***join(): joining on index***


combines DataFrames using their indexes by default. It is useful when the matching information is stored in row labels rather than regular columns.

```python
data1 = pd.DataFrame({
    'City': ['Pune', 'Mumbai', 'Delhi']
}, index=['Anna', 'Ben', 'Chris'])

data2 = pd.DataFrame({
    'Lemonades': [2, 3, 4]
}, index=['Anna', 'Ben', 'Chris'])

result = data1.join(data2)
print(result)
```
