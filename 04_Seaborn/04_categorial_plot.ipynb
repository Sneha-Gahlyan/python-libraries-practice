# 04. Categorical Plots

Categorical plots are used to visualize and analyze data that is divided into different categories or groups.

Categorical plots ka use hum tab karte hain jab humein different categories ke beech comparison karna hota hai.

# Categorical Scatter Plot

###* Stripplot : axes level function

graph btn numerical and categorical data

```python
import seaborn as sns
```

```python
# import dataset

iris = sns.load_dataset('iris')
tips = sns.load_dataset('tips')
titanic = sns.load_dataset('titanic')
#
```

```python
sns.stripplot(x = 'day', y = 'total_bill', data = tips)
```

```python
# for one single line

sns.stripplot(x = 'day', y = 'total_bill', data = tips, jitter = False)
```

#Figure level function


###catplot

```python
sns.catplot(x = 'day', y = 'total_bill', data = tips, kind = 'strip')
```

```python
#jitter

sns.catplot(x = 'day', y = 'total_bill', data = tips, kind = 'strip', jitter = 0.2)
```

```python
#hue

sns.catplot(x = 'day', y = 'total_bill', data = tips, kind = 'strip', hue = 'sex')
```

#* Swarm plot

```python
sns.swarmplot(x = 'day', y = 'total_bill', data = tips)
```

```python
sns.catplot(x = 'day', y = 'total_bill', data = tips, kind = 'swarm')
#
```

#-> categorical distribution plot

### *boxplot : axes level


A boxplot is a stanardized way of displaying the distribution of data based on five number summary('minimum',first quartile[Q1],median,third quartile[Q3] and 'maximum'). It can tell you about outliers and what their values are.

Boxplots can also tell you if your data iis symmetrical, how tightly your data is grouped and if and how your data is skewed.

```python
sns.boxplot(data=tips , x = 'sex' , y='total_bill')
```

```python
# single boxplot -> numerical col

sns.boxplot(data=tips , y='total_bill')
```

```python
sns.catplot(x = 'day', y = 'total_bill', data = tips, kind = 'box')
```

```python
#hue

sns.catplot(x = 'day', y = 'total_bill', data = tips, kind = 'box', hue = 'sex')
```

# * violinplot

```python
sns.violinplot(data=tips , x = 'day' , y='total_bill')
```

```python
#catplot and hue

sns.catplot(x = 'day', y = 'total_bill', data = tips, kind = 'violin' , hue = 'sex')
```

```python
# split

sns.catplot(x = 'day', y = 'total_bill', data = tips, kind = 'violin' , hue = 'sex' , split = True)
```

#Categorical estimation plot

### * bar plot


An error bar is a line shown above and/or below a plotted value to represent the uncertainty or variability in the data.

In simple words, an error bar helps us understand how much the actual data values may vary around the estimated value (such as the mean).

```python
### * bar plot

sns.barplot(data=tips , x = 'sex' , y='total_bill')
```

```python
sns.barplot(data=tips , x = 'sex' , y='total_bill' , hue = 'smoker' )
```

```python
#estimator

import numpy as np

sns.barplot(data=tips , x = 'sex' , y='total_bill' , hue = 'smoker' , estimator = np.min)
```

### * point plot

```python
sns.pointplot(data=tips , x = 'sex' , y='total_bill' , hue = 'smoker', estimator=np.min )
```

#faceting

```python
sns.catplot(data=tips , x = 'sex' , y = 'total_bill' , kind = 'box', col = 'smoker')
#
```

## * count plot

```python
sns.countplot(data=tips , x = 'sex'  , hue = 'day')
```
