```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import pandas as pd
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

- Scatter plot : axes level function

```python
sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip')
```

```python
sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip' , hue = 'sex')
```

```python
sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip' , hue = 'sex' , style = 'time')
```

```python
sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip' , hue = 'sex' , style = 'time' , size = 'size')
```

- Scatter plot : figure level function

```python
# relplot : fig level fn -> square shape

# all styling of scatter are also available in this...

sns.relplot(data = tips , x = 'total_bill' , y = 'tip' , kind = 'scatter')
```

#Line plot

```python
gap = px.data.gapminder()

temp_df = gap[gap['country'] == 'India']
```

```python
temp_df
```

#Axes level function

```python
sns.lineplot(data = temp_df , x = 'year' , y = 'lifeExp')
```

```python
#hue style

temp_df2 = gap[gap['country'].isin(['India' , 'China' , 'United States'])]

sns.lineplot(data = temp_df2 , x = 'year' , y = 'lifeExp' , hue = 'country')
```

#Figure level function

```python
# using relplot

sns.relplot(data = temp_df , x = 'year' , y = 'lifeExp' , kind = 'line')
```

```python
#hue style

sns.relplot(data = temp_df2 , x = 'year' , y = 'lifeExp' , hue = 'country' , kind = 'line')
```

```python
sns.relplot(data = temp_df2 , x = 'year' , y = 'lifeExp' , hue = 'country' , kind = 'line' , style = 'continent' , size = 'continent')
```

#Facet plot


- figure level function -> work with relplot

- it will not work with scatter plot and lineplot

```python
sns.relplot( data = tips , x = 'total_bill' , y = 'tip' , col = 'smoker' , row = 'sex' , kind = 'scatter')
#
```

#col wrap

```python
sns.relplot( data = gap , x = 'lifeExp' , y = 'gdpPercap' , col = 'year' , kind = 'scatter' , col_wrap = 2)
```
