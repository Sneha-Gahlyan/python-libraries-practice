```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import pandas as pd
```

#Heatmap : axes level fn

Definition: A Heatmap is a data visualization technique that represents numerical values using different colors. It helps us identify patterns, relationships and high or low values in a dataset.


plot rectangular data as a color encoded matrix

```python
gap = px.data.gapminder()

gap.head()
```

```python
temp_df = gap.pivot(index='country', columns = 'year' , values = 'lifeExp')

# use figure in axes fn

plt.figure(figsize = (10,10))
sns.heatmap(temp_df)
```

```python
#annot

temp_df2 = gap[gap['continent'] == 'Europe'].pivot(index='country', columns = 'year' , values = 'lifeExp')

plt.figure(figsize = (10,10))
sns.heatmap(temp_df2, annot = True , linewidths=0.2 , cmap = 'viridis')
#
```

#Clustermap

A Clustermap is a data visualization technique used to display numerical data as a heatmap along with hierarchical clustering. It groups similar rows and columns together based on their values.



Plot a matrix dataset as a heirarchically - clustered heatmap

this fn requires scipy to be available

```python
data = {
    'Maths': [90, 85, 20, 25, 88],
    'Science': [95, 80, 30, 20, 90],
    'English': [85, 90, 25, 30, 87],
    'Computer': [92, 88, 15, 25, 95]
}

df = pd.DataFrame(
    data,
    index=['A', 'B', 'C', 'D', 'E']
)

print(df)
```

```python
sns.clustermap(df, cmap='coolwarm', annot=True)

plt.show()
```
