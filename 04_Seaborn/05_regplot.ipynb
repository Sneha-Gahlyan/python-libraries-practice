#Regression:
Regression is a statistical technique used to understand the relationship between two numerical variables and predict how one variable changes when another variable changes.

```python
import seaborn as sns
```

```python
# import dataset

iris = sns.load_dataset('iris')
tips = sns.load_dataset('tips')
titanic = sns.load_dataset('titanic')
#
```

# *Regression Plot
A Regression Plot is a visualization that displays:
- Individual data points using dots.
- A regression line (best-fit line).
- Sometimes a confidence interval around the regression line.

* regplot() in Seaborn?

regplot() is a Seaborn function used to create a scatter plot with a fitted regression line

```python
# axes level fn
sns.regplot(x='total_bill', y='tip', data=tips)
```

```python
# formula
sns.lmplot(x='total_bill', y='tip', data=tips , hue = 'sex')
#
```

# *residual plot


```python
sns.residplot(data = tips , x = 'total_bill' , y = 'tip')
```
