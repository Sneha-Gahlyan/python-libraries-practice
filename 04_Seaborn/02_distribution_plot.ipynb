#Distribution plot

Distribution plots are used to understand how numerical data is distributed. They help us identify the frequency, spread, shape, skewness and patterns in a dataset.

- used for univariate analysis

- used to find out the distribution

- Range of the observation

- Central Tendancy

- Is the data bimodal?

- Are there outliers?




the main distribution plotting functions are:
1. Histogram (histplot)
2. Kernel Density Estimate (kdeplot)
3. Rug Plot (rugplot)

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import pandas as pd
```

### figure level -> displot


### axes level -> hisplot,kdeplot,rugplot

```python
tips = sns.load_dataset('tips')
```

#Histogram


A Histogram is a data visualization technique used to represent the frequency distribution of numerical data using rectangular bars.

* kisi numerical column mein kaunsi value range kitni baar aa rahi hai.

```python
sns.histplot(data=tips,x='total_bill')
```

#Figure level function -> displot

```python
sns.displot(data=tips,x='total_bill' , kind = 'hist')
```

```python
#bins parameter

sns.displot(data=tips,x='total_bill' , kind = 'hist',bins=20)
```

* It's also possible to visualise the distribution of a categorical variable using the logic of a histogram

* discrete bins are automatically set for categorical variables

```python
sns.displot(data=tips , x = 'day' , kind = 'hist')
```

```python
#hue parameter

sns.displot(data=tips , x = 'tip' , kind = 'hist',hue='sex')
```

```python
# element step

sns.displot(data=tips , x = 'tip' , kind = 'hist',hue='sex',element='step')
```

```python
titanic = sns.load_dataset('titanic')

titanic.head()
```

```python
sns.displot(data=titanic , x = 'age' , kind = 'hist' , element = 'step' , hue = 'sex')
```

###faceting using row and col

#Kde plot


A KDE plot smooths the observation with a gaussian kernel , producing a continuous density estimates

```python
sns.kdeplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='kde')
```

```python
# hue - > fill

sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill = True)
```

#Rugplot

plot marginal distribution by drawing ticks along wit x and y axes

This fn is invented to comlement other plots by showing th location of individual observations



```python
sns.kdeplot(data=tips,x='total_bill')

sns.rugplot(data=tips,x='total_bill')
```

#bi- variate hisogram


histogram bins the data within rectangles that tile the plot
and then shows the count of observations within each rectangle with the fill color

```python
sns.histplot(data=tips,x='total_bill',y='tip')
```

#bivariate kdeplot

smoth the x,y observation with a 2d guassian

```python
sns.kdeplot(data=tips,x='total_bill',y='tip')
```
