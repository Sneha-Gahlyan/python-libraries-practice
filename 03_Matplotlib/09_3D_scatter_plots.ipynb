#3D Scatter plots

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
batters = pd.read_csv('/content/batter.csv')

batters.head()
```

```python
fig = plt.figure()

ax = plt.subplot(projection = '3d')

ax.scatter3D(batters['runs'] , batters['avg'] , batters['strike_rate'])

ax.set_title('IPL Btsman analysis')
ax.set_xlabel('runs')
ax.set_ylabel('avg')
ax.set_zlabel('strike_rate')

plt.show()
#
```

#3D surface plot

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx , yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2

z.shape
```

```python
fig = plt.figure(figsize=(12,9))

ax = plt.subplot(projection = '3d')

p = ax.plot_surface(xx,yy,z, cmap = 'viridis')

fig.colorbar(p)

plt.show()
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,9))

ax = plt.subplot(projection = '3d')

p = ax.plot_surface(xx,yy,z, cmap = 'viridis')

fig.colorbar(p)

plt.show()
```

#Contour plots

it replresent 3d graph in 2d

```python


fig = plt.figure(figsize=(12,9))

ax = plt.subplot()

p = ax.contour(xx,yy,z, cmap = 'viridis')

fig.colorbar(p)

plt.show()
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,9))

ax = plt.subplot()

p = ax.contourf(xx,yy,z, cmap = 'viridis')

fig.colorbar(p)

plt.show()
```
