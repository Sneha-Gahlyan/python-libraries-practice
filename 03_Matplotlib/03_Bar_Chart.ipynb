#Stacked bar chart

```python
plt.bar(df['batsman'],df['2015'],color='red' , label = '2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'], label = '2016' , color='blue')
plt.bar(df['batsman'],df['2017'],bottom=df['2015']+df['2016'], label = '2017',color='green')

plt.legend()

plt.xlabel('Batsman')
plt.ylabel('Total Runs')
plt.title('Batsman Runs by Season')

```

2015 bar starts at 0 and reaches 513.



2016 bar starts at 513 and reaches 1200.



2017 bar starts at 1200 and reaches 1416.
