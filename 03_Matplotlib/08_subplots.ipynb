#Subplots

a diff way to plot  graphs

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
batters = pd.read_csv('/content/batter.csv')
batters.head()
```

```python
plt.scatter(batters['avg'] , batters['strike_rate'])
```

###Second way of plot subplot

```python
fig , ax = plt.subplots(figsize=(10,5))

ax.scatter(batters['avg'] , batters['strike_rate'])

ax.set_title('Scatter Plot')
ax.set_xlabel('Average')
ax.set_ylabel('Strike Rate')

plt.show()
```

```python
plt.subplot()
```

### For ploting different plots

```python
plt.subplots(nrows = 2 , ncols = 2)

plt.show()
```

```python
fig , ax = plt.subplots(nrows = 2 , ncols = 1 , figsize=(10,5) , sharex = True)

ax[0].scatter(batters['avg'] , batters['strike_rate'])
ax[0].set_title('Avg vs Scatter Rate')

ax[0].set_ylabel('Strike Rate')


ax[1].scatter(batters['avg'] , batters['runs'])
ax[1].set_title('Avg vs runs')
ax[1].set_xlabel('Average')
ax[1].set_ylabel('Runs' )

plt.show()
```

```python
fig , ax = plt.subplots(nrows = 2 , ncols = 2 , figsize=(10,5) )

ax[0,0].scatter(batters['avg'] , batters['strike_rate'])
ax[0,0].set_title('Avg vs Scatter Rate')
ax[0,0].set_ylabel('Strike Rate')


ax[0,1].scatter(batters['avg'] , batters['runs'])
ax[0,1].set_title('Avg vs runs')
ax[0,1].set_xlabel('Average')
ax[0,1].set_ylabel('Runs' )

ax[1,0].hist(batters['avg'])

ax[1,1].hist(batters['runs'])

plt.show()
```

```python
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batters['avg'],color='green')

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters['runs'],color='blue')

plt.show()
```
