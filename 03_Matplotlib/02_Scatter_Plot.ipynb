#Scatter Plot

A Scatter Plot is a graph used to show the relationship between two numerical variables using individual dots (data points).

Unlike a line plot, a scatter plot does not connect the points with lines.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = np.linspace(-10,10,50)

y = 10*x + 3 + np.random.randint(0 , 300 , 50)
```

```python
plt.scatter(x,y)
```

#plt.scatter on pandas data

```python
df = pd.read_csv("/content/batter.csv")
df
```

```python
plt.scatter(df['avg'] , df['strike_rate'],marker = "+")
plt.title('Avg and SR analysis of top 50 Batsman')

plt.xlabel('Average')
plt.ylabel('Strike Rate')
```

#Seaborn data set


```python
tips = sns.load_dataset('tips')

tips
```

```python
#slower
plt.scatter(tips['total_bill'] , tips['tip'])
```

```python
plt.scatter(tips['total_bill'] , tips['tip'] , s = tips['size']*20)
```

```python
# scatter plot using pllt.plot
#faster
plt.plot(tips['total_bill'],tips['tip'],'o')
```
