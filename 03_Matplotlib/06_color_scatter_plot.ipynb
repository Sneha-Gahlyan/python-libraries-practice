#Scatter plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris = pd.read_csv('/content/iris.csv')

iris.head()
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'])

plt.xlabel('Sepal Length Cm')
plt.ylabel('Petal Length Cm')
plt.title('Scatter Plot')

plt.show()
```

###Replacing

iris.sample() is a Pandas function used to randomly select rows from the iris DataFrame.

```python
iris['Species'] = iris['Species'].replace('Iris-setosa', 0)
iris['Species'] = iris['Species'].replace('Iris-versicolor', 1)
iris['Species'] = iris['Species'].replace('Iris-virginica', 2)

iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c= iris['Species'])

plt.xlabel('Sepal Length Cm')
plt.ylabel('Petal Length Cm')
plt.title('Scatter Plot')

plt.show()
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'] , c= iris['Species'] , cmap = 'winter')

plt.xlabel('Sepal Length Cm')
plt.ylabel('Petal Length Cm')
plt.title('Scatter Plot')

plt.show()
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'] , c= iris['Species'] , cmap = 'winter')

plt.xlabel('Sepal Length Cm')
plt.ylabel('Petal Length Cm')
plt.title('Scatter Plot')

plt.colorbar()
plt.show()
```

#Alpha :

for opacity

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'] , c= iris['Species'] , cmap = 'winter', alpha = 0.6)

plt.xlabel('Sepal Length Cm')
plt.ylabel('Petal Length Cm')
plt.title('Scatter Plot')
plt.colorbar()
plt.show()
```

#plot size



```python
plt.figure(figsize = (12, 8))

plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'] , c= iris['Species'] , cmap = 'winter', alpha = 0.6)

plt.xlabel('Sepal Length Cm')
plt.ylabel('Petal Length Cm')
plt.title('Scatter Plot')
plt.colorbar()
plt.show()
```
