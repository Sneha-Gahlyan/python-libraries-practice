#***Pie Chart***

A Pie Chart is a circular chart divided into slices. Each slice represents a category's contribution to the total.

 A pie chart shows how a total is divided into different parts or percentages.

### website
matplotlib.org

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

- Univariate/Bivariate

- Categorial vs numerical

- Usse case : To find contribution on a standard scale

```python
# simple data

data = [68,88,93,75,44]
subjects = ['eng' , 'science' , 'maths' , 'sst' , 'hindi']

plt.pie(data,labels=subjects)
plt.show()
```

```python
#dataset

df = pd.read_csv('/content/gayle-175.csv')

df
```

```python
plt.pie(df['batsman_runs'] , labels=df['batsman'] , autopct = '%0.1f%%')
plt.show()
```

```python
plt.pie(df['batsman_runs'] , labels=df['batsman'] , autopct = '%0.1f%%' , colors = ['green' , 'pink' , 'yellow' , 'red' , 'teal' , 'brown'])
plt.show()
```

```python
# can be extract particular area

plt.pie(df['batsman_runs'] , labels=df['batsman'] , explode = [0.2 , 0,0,0,0,0] ,autopct = '%0.1f%%' , colors = ['green' , 'pink' , 'yellow' , 'red' , 'teal' , 'brown'])

plt.show()
```

```python
#Shadow

plt.pie(df['batsman_runs'] , labels=df['batsman'] ,explode = [0.2,0,0,0,0,0], shadow=True, autopct = '%0.1f%%' , colors = ['green' , 'pink' , 'yellow' , 'red' , 'teal' , 'brown'])
plt.show()
```
