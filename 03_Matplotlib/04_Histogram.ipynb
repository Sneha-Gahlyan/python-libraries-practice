#Histogram

- For numerical column
- Univariate Analysis
- Use case : Frequency Count

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = [3,6,90,2,4,32]
plt.hist(data)
```

#Bins

```python
data2 = [32,45,10,66,72,89]
plt.hist(data2 , bins = [10,25,30,45,50,60,70])

plt.show()
```

#Logarithmic scale

A logarithmic scale (log scale) is used when data values have a very large difference between them.

```python
runs = [10, 20, 30333, 50, 1000, 200, 500, 100000, 5000, 10000]

plt.hist(runs, bins=[10,20,30,40,50,60 ], log = True)

plt.show()
```
