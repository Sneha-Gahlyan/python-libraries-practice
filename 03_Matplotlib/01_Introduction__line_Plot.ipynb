#Matplotlib in Python?

Matplotlib is a Python library used for data visualization. It helps us represent data in the form of graphs, charts, and plots so that we can understand and analyze data easily.

#Types of Data

1. Numerical data
2. Categorial data

###1. Univariate Analysis (1 Variable)

Univariate analysis means analyzing only one column at a time.

###2. Bivariate Analysis (2 Variables)

Bivariate analysis means analyzing two cs together to understand the relationship or comparison between them.

###3. Multivariate Analysis (3 or More Variables)

Multivariate analysis means analyzing three or more variables together to understand relationships and patterns between them.

```python
#marker


plt.plot(batsman['index'], batsman["V Kohli"],color = 'Green' , linestyle = 'solid',marker = 'D',markersize=10)
plt.plot(batsman['index'], batsman["RG Sharma"], color = "Red" , linestyle = 'dashed',marker='D')

plt.title('Rohit sharma Vs Virat kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored ')
```

#Label

```python
plt.plot(batsman['index'], batsman["V Kohli"],color = 'Green' , linestyle = 'solid',marker = 'D',markersize=10 , label= "Rohit")
plt.plot(batsman['index'], batsman["RG Sharma"], color = "Red" , linestyle = 'dashed',marker='D' , label = 'Virat')

plt.title('Rohit sharma Vs Virat kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored ')

plt.legend()
#plt.legend(loc = 'best')
```

#Limiting axes

```python
hours = [1, 2, 3, 4, 5,6]
marks = [20, 40, 50, 70, 90,800000]

plt.plot(hours, marks)
```

Limiting axes means controlling the range of values displayed on the X-axis and Y-axis of a graph.

In simple words, sometimes your graph displays a very large range of values, but you only want to focus on a specific portion. You can limit the axes to show only that portion.


Function	Purpose

plt.xlim()	Set X-axis limits

plt.ylim()	Set Y-axis limits

plt.axis()	Set both X and Y limits

plt.axis('auto')	Automatically adjust axis limits

```python
hours = [1, 2, 3, 4, 5,6]
marks = [20, 40, 50, 70, 90,800000]

plt.plot(hours, marks)

#limitng axis
plt.ylim(0,400)
```

#Grid in Matplotlib?

Grid is used to display horizontal and vertical reference lines in a graph.

These lines help us read and compare values easily by looking at the X-axis and Y-axis.

```python
hours = [1, 2, 3, 4, 5,6]
marks = [20, 40, 50, 70, 90,80]

plt.plot(hours, marks)

plt.grid()
```
