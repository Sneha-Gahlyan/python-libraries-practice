#Pandas Plot()

using pandas plot() build matplotlib graph

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# on a series

s = pd.Series([1,2,4,5,8,9])

s.plot(kind = 'pie')
```

#can be used on a dataframe as well

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
#scatter plot

tips.plot(kind = 'scatter' , x = 'total_bill' , y = 'tip')
```

```python
tips['size'] = tips['size']*100
```

```python
#scatter plot

tips.plot(kind = 'scatter' , x = 'total_bill' , y = 'tip' , title = 'cost analysis' , c = 'sex', cmap = 'viridis',marker = '+' , figsize = (10,6) , s = 'size' )
```

#Line plot

```python
import pandas as pd

stocks = pd.DataFrame({
    'Date': ['2021-05-24', '2021-05-31', '2021-06-07',
             '2021-06-14', '2021-06-21'],

    'MSFT': [249.679993, 250.789993, 257.890015,
             259.429993, 265.019989],

    'FB': [328.730011, 330.350006, 331.260010,
           329.660004, 341.369995],

    'AAPL': [124.610001, 125.889999, 127.349998,
             130.460007, 133.110001]
})

stocks
```

```python
stocks['MSFT'].plot(kind = 'line')
```

```python
stocks.plot(kind = 'line' , x = 'Date')
```

```python
stocks[['Date' , 'AAPL' , 'FB']].plot(kind = 'line' , x = 'Date')
```

#Histogram

```python
stocks.plot(kind = 'hist')
```

#Pie chart

```python
stocks['MSFT'].plot(kind = 'pie' , labels = stocks['Date'].values , autopct = '%0.1f%%')
```

```python
#multilple pie chart


stocks[['MSFT' , 'FB' , 'AAPL']].plot(kind='pie', subplots=True, figsize=(8, 8))
plt.show()
```

#Pie chart on multiindex dataframes

```python
tips.pivot_table(index = 'day' , columns = 'sex' , values = 'total_bill' , aggfunc = 'mean').plot(kind = 'line')
```

```python
tips.pivot_table(index = ['day' , 'time'] , columns = ['sex','smoker'] , values = 'total_bill' , aggfunc = 'mean').plot(kind = 'line')
```
