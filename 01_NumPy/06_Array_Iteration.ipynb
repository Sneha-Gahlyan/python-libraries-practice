#Array Iteration

###Iterating over a 1D array

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

for x in arr:
    print(x)
```

###Iterating over a 2D array

```python
arr2 = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

for row in arr2:
    print(row)

#By default, iterating over a 2D array gives you one row at a time.

#To access every individual element, use a nested loop:

for row in arr2:
    for x in row:
        print(x)
```

###Iterating over a 3D array

```python
arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

for layer in arr3:
    for row in layer:
        for x in row:
            print(x)
```

###np.nditer()
NumPy provides np.nditer() to iterate through every element of an array without writing nested loops

```python
# 2D

print(arr2)
print()
for x in np.nditer(arr2):
    print(x)


#flags=['buffered']
#The buffered flag allows NumPy to use a temporary buffer when necessary, such as when converting data types during iteration.

#op_dtypes=['S']
#op_dtypes specifies the desired data type for the iterator's output.

print()
#3D
for i in np.nditer(
    arr3,
    flags=['buffered'],
    op_dtypes=['S']
):
    print(i)
```

###np.ndenumerate()
If you need both the index and the value, use np.ndenumerate()

```python
arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
for index, value in np.ndenumerate(arr3):
    print(index, value)
```

#The difference between copy and view

```python
# copy
# Creates a separate array with its own data. Changes to the copy do not affect the original array.
a = np.array([1,2,3,4])

b = a.copy()

print("a : ",a)
print("b : ",b)

print()

# if do change in original it will not shown in copy
a[2] = 89
print("a : ",a)
print("b : ",b)
```

```python
#view
#- Creates a new way to access the same underlying data. Changes to the view can affect the original array.

x = np.array([9,8,4,6,5,4])

y = x.view()

print("x: ",x)
print("y: ",y)


print()
# if there any change in original it will see in view
x[2] = 32
print("x: ",x)
print("y: ",y)
```

#Join Array

Joining arrays means combining two or more arrays into a single array.

### np.concatenate()
This is the most common method for joining arrays.

```python
import numpy as np

# joining 1D array
arr1 = np.array([10, 20, 30])
arr2 = np.array([40, 50, 60])

arr = np.concatenate((arr1, arr2))

print(arr)
```

```python
# Joining 2D array
arr1 = np.array([[1, 2],
                 [3, 4]])

arr2 = np.array([[5, 6],
                 [7, 8]])

print(np.concatenate((arr1, arr2), axis=0))
```

```python
import numpy as np

arr1 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

arr2 = np.array([
    [[9, 10], [11, 12]],
    [[13, 14], [15, 16]]
])

print(arr1.shape)  # (2, 2, 2)

result = np.concatenate((arr1, arr2), axis=2)
print(result)
```

###np.stack

```python
arr3 = np.array([1, 2, 3])
arr4 = np.array([4, 5, 6])

print("Stack")
print(np.stack((arr3, arr4)))
print("Axis = 1")
print(np.stack((arr3, arr4),axis =1))

#hstack
print("hstack:")  #row
print(np.hstack((arr3 , arr4)))

print("vstack")   #column
print(np.vstack((arr3,arr4)))

print("dstack")   #height
print(np.dstack((arr3,arr4)))
```
