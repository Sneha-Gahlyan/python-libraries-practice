#Data Types in NumPy array


NumPy arrays store elements using a specific data type, known as dtype.

```python
import numpy as np

arr1 = np.array([10, 20, 30])
print("Data Type of arr1",arr1.dtype)

arr2 = np.array([10.5, 20.5, 30.5])
print("Data Type of arr2",arr2.dtype)

arr3 = np.array([True, False, True])
print("Data Type of arr3",arr3.dtype)

arr4 = np.array(["Python", "NumPy"])
print("Data Type of arr4 ",arr4.dtype)
```

### Specifying Data Types

We can specify the desired data type while creating an array by using the `dtype` parameter.

**Syntax:** `np.array(data, dtype=type)`

```python
x = np.array([1,2,34,5] , dtype = np.int8)

print(x)
print(x.dtype)
```

```python
# using function
x = np.array([1,2,34,5] , dtype = "f")

print(x)
print(x.dtype)
```

### Data Types as function


```python

x1 = np.array([1,2,3,4])

x2  = np.float32(x1)

x3 = np.int_(x2)

print(f"X1 :",x1 , x1.dtype)
print(f"X2 :",x2 , x2.dtype)
print(f"X3 :",x3 , x3.dtype)
```

#Shape and reshaping in NumPy arrays

###Shape

```python
# Shape : In NumPy, shape tells us how many elements an array has along each dimension

import numpy as np

# 1D array
arr1 = np.array([10, 20, 30, 40])
print(arr1.shape)

# 2D array
arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])
print(arr2.shape)

# 3D array
arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print(arr3.shape)
```

```python
# for n dimension

arrn = np.array([1,2,3,4], ndmin = 4)

print(arrn)
print(arrn.ndim)
print()
print(arrn.shape)
```

###Reshape

The reshape() method changes the shape of an array without changing its elements or their order.

Syntax: array.reshape(rows, columns)

Important: The new shape must have the same total number of elements as the original array.

```python
var = np.array([1,2,3,4,5,6])
print("Dimension of var: " , var.ndim)
print(var)

new = var.reshape(3,2)
print("Dimension of new: ",new.ndim)

print(new)
```

```python
#3-Dimension

var3 = np.array([1,2,3,4,5,6,4,8,9,10,11,12])

print(var3)

nvar = var3.reshape(2,3,2)
print(nvar)

print(nvar.ndim)


#n-D to 1-D  : by using -1 in reshape

nD = nvar.reshape(-1)
print(nD)
print(nD.ndim)
```

## 4. Using `-1` in `reshape()`

NumPy can automatically calculate one dimension when we use `-1`.

For example, `reshape(2, -1)` means: create 2 rows and let NumPy determine the number of columns.

Only one dimension can be specified as `-1` in a single reshape operation.

```python
arr = np.arange(1, 13)

print(arr.reshape(2, -1))

print(arr.reshape(3, -1))

print(arr.reshape(-1, 4))
```
