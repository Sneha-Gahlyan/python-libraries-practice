# Difference Between NumPy Array and Python List

## 1. Data Type Storage
- Python lists can store different data types.
- NumPy arrays generally store elements of the same data type.

## 2. Importing Module
- Python lists are built into Python; no import is required.
- NumPy requires importing the library using `import numpy as np`.

## 3. Numerical Operations
- Python lists do not perform element-wise arithmetic directly.
- NumPy arrays support element-wise mathematical operations.

## 4. Modification Capabilities
- lists have less method of modification
- NumPy have large method of modification

## 5. Memory Usage
- NumPy arrays often use less memory for large, homogeneous numerical data.

## 6. Speed
- NumPy operations are often faster for large numerical calculations.

## 7. Convenience
- NumPy provides built-in mathematical and statistical functions.

#Special NumPy array

###1. Array filled with 0's

```python
# 1-D
ar_zero = np.zeros(4)

print(ar_zero)

print()

# 2-D
ar_zero_2 = np.zeros((3,4))

print(ar_zero_2)
```

###2. Array filled with 1's

```python
# 1-D
ar_one = np.ones(4)
print(ar_one)

print()

# 2-D
ar_one_2 = np.ones((3,4))
print(ar_one_2)
```

###3. Create an empty array

```python
ar_empty = np.empty(4)
print(ar_empty)
```

###4. An array with a range of elements

```python
ar_range = np.arange(4)
print(ar_range)
```

###5. Array diagonal element filled with 1's

```python
ar_dia = np.eye(3)
print(ar_dia)
```

```python
ar_diagonal = np.eye(3,5)
print(ar_diagonal)
```

###6. Create an array with values that are spaced linearly in a specified interval

```python
ar_lin = np.linspace(1,10,num = 5)
print(ar_lin)
```

# Create NumPy array with random numbers

#rand()

The function is used to generate  a random value between o and 1.

```python
# 1-D
var = np.random.rand(4)
print(var)
```

```python
# n-D
var1 = np.random.rand(3,5)
print(var1)
```

#randn()

The function is used to generate a random value close to zero. This may return positive and negative number as well.

```python
var2 = np.random.randn(5)
print(var2)
```

#randf()

The function for doing random sampling in numpy. It returns an array of specified shape and filled it with random floats in half-open interval [0.0,1.0)

```python
var3 = np.random.ranf(4)
print(var3)
```

#randint()

The function is used to generate a random Number between a given range.

```python
# syntax : np.random.randint(min,max,total_values)

var4 = np.random.randint(5,20,5)
print(var4)
```
