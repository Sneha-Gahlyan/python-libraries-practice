#Introduction to NumPy

#What is NumPy?

NumPy is the fundamental package for scientific computing in Python.
NumPy is a Python library that provides a multidimensional array object, various
derived objects

It provides fast and efficient operations on large amounts of numerical data.

#NumPy vs Python list

###Advantages of using Numpy Arrays Over Python Lists:

- consumes less memory.
- fast as compared to the python List.
- convenient to use.

# Create a Python list and NumPy array

```python
import numpy as np
# python list

l = [1,2,3,4]

# numpy array

arr = np.array([1,2,3,4])


print("Python list : ",l)
print("Numpy array : ",arr)
```

###Basic numerical operations



```python
numbers = np.array([10, 20, 30, 40])

print("Original Array:", numbers)
print("Addition:", numbers + 5)
print("Multiplication:", numbers * 2)
print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
```
