#Arithmetic Operation in NumPy array

#1-Dimension array

```python
import numpy as np

# Addition of an array
var1  = np.array([1,2,3,4])

varadd1 = var1 + 3
print( "Addition of an array: ",varadd1)

# Subtraction of an array
varadd2 = var1 - 3
print( "Subtraction of an array: ",varadd2)

# Multiplication of an array
varadd3 = var1 * 3
print( "Multiplication of an array: ",varadd3)

# Division of an array
varadd4 = var1 / 3
print("Division of an array: ",varadd4)

#Modulus of an array'
varadd5 = var1 % 3
print( "Modulus of an array: ",varadd5)
```

```python
# Arithmetic Operation on two array

var2 = np.array([1, 2, 3, 4])
var3 = np.array([1, 2, 3, 4])

addition1 = var2 + var3

print(addition1)
```

###Arithmetic operation using function

```python
arr1 = np.array([1,4,6,9])
arr2 = np.array([4,8,9,10])

#addition
addition2 = np.add(arr1,arr2)
print("Addition: ",addition2)

#Subtraction
sub = np.subtract(arr1,arr2)
print("Subtraction: ",sub)

#Multiplication
mul = np.multiply(arr1,arr2)
print("Multiplication: ",mul)

#Division
div = np.divide(arr1,arr2)
print("Division: ",div)

#Mod
mod = np.mod(arr1,arr2)
print("Mod: ",mod)

#Power
power = np.power(arr1,arr2)
print("Power: ",power)

#Reciprocal
rec = np.reciprocal(arr1)
print("Reciprocal: ",rec)

```

#2-Dimension array

```python
arr3 = np.array([[1, 2, 3],
                 [4, 5, 6]])

arr4 = np.array([[10, 20, 30],
                 [40, 50, 60]])

print(arr3)
print()
print(arr4)
print()

#Addition
add = arr3 + arr4
print(add)

print()

#Multiplication
multiply = arr3 * arr4
print(multiply)
```

#Basic Arithmetic Operation

```python
# min and max according to 1-Dimension
arr5  = np.array([1,2,4,6,0])

print(arr5)

#np.argmin(x) : Returns the index (position) of the smallest element in an array.
min = np.min(arr5)
print("Minimum : ",min," at index : ",np.argmin(arr5))


max = np.max(arr5)
print("Maximum : ",max," at index : " , np.argmax(arr5))
```

```python
import numpy as np

#min and max according to 2-Dimension
arr6 = np.array([[1,2,10],
                  [4,5,6]])


# along column : axis = 0

#min
print("Min value acc to column: ",np.min(arr6,axis = 0))

#max
print("Max value acc to column: ",np.max(arr6,axis = 0))


#along row : axis = 1

#min
print("Min value acc to row: ",np.min(arr6,axis = 1))

#max
print("Max value acc to row: ",np.max(arr6,axis = 1))
```

###np.sqrt(x) : Calculates the square root of each element in an array.

```python
print(np.sqrt(arr6))
```

###sin and cos value

```python
print("sin value: ",np.sin(arr6))

print()

print("cos value: ",np.cos(arr6))
```

###np.cumsum(x) : Calculates the cumulative sum — the running total of array elements.

```python
print(np.cumsum(arr6))

n = np.cumsum(arr6)
print(n.ndim)
```

#Broadcasting with NumPy arrays

###What is broadcasting?
 Broadcasting is a NumPy feature that allows arithmetic operations between arrays of different shapes, provided their shapes are compatible.

```python
# Two arrays of different shapes

arr7 = np.array([10, 20, 30]) # 1*3
arr8 = np.array([1]) #1*1

print(arr7 + arr8) #1*3

#Here, arr7 has shape (1,3), while arr8 has shape (1,). NumPy can broadcast the single value 1 across the three elements.
```

###Broadcasting with 2-D array

```python
arr1 = np.array([[1, 2, 3],
                 [4, 5, 6]]) #2*3

arr2 = np.array([10, 20, 30]) #1*3

print(arr1 + arr2) #2*3
```

###. Example of incompatible shapes

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

b = np.array([10, 20])

print(a + b)

#This raises a ValueError because the trailing dimensions 3 and 2 are neither equal nor 1.
```
