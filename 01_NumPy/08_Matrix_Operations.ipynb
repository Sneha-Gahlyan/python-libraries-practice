#Matrix

```python
import numpy as np

var = np.matrix([[1,2,3], [1,2,3]])
print(var)
print(type(var))
```

```python
var1 = np.array([[1,2,3], [1,2,3]])
print(var1)
print(type(var1))
```

```python
var2 = np.matrix([[1,2], [1,2]])
var3 = np.matrix([[1,2], [1,2]])

print(var)
print(type(var))

print()
#Multiplication
print(var2 * var3)

#Multiplication using dot function
print()
print(var2.dot(var3))
```

#Matrix function

1. Transpose

 Transpose changes rows into columns and columns into rows.

```python
var4 = np.matrix([[1,2,3], [4,5,6]])
print(var4)
print()
print(np.transpose(var4))

print()
print(var4.T)
```

2. Swapaxes

 It swaps two specified axes of an array.

For a 2D matrix, axis=0 represents rows and axis=1 represents columns

```python
#np.swapaxes(array, axis1, axis2)

var5 = np.array([[1, 2, 3],
                 [4, 5, 6]])

print(np.swapaxes(var5 , 0, 1))
```

```python

var6 = np.array([
    [[1, 2],
     [3, 4]],

    [[5, 6],
     [7, 8]]
])

print(var6.shape)

print(np.swapaxes(var6, 0, 1))
```

3. Inverse

```python
var7 = np.matrix([[1,2], [3,4]])
print(var7)
print()
print(np.linalg.inv(var7))
```

4. Power

```python
#n>0
print(np.linalg.matrix_power(var7 , 2))

#n=0
print(np.linalg.matrix_power(var7 , 0))

#n<0
#inverse * power
print(np.linalg.matrix_power(var7 , -2))
```

5. Determinate

```python
var8 = np.matrix([[1,2,3],[3,4,3], [1,2,3]])
print(var8)
print()
print(np.linalg.det(var8))
```
