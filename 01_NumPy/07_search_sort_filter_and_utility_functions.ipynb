#Search Array

Search an array for a certain value, and return
the indexes that get a match.

```python
import numpy as  np

arr = np.array([1,2,3,4,2,5,6,8])

x = np.where( arr == 2)
y = np.where( (arr % 2) == 0 )
print(x)
print(y)
```

#Search Sorted Array

###-Find the insertion position
which performs a binary search in the
array, and returns the index where the specified value would be
inserted to maintain the search order.

Important: arr2 have to be sorted. np.searchsorted() assumes the array is sorted by default,

```python
#side="left" — Insert before existing 20s
#side="right" — Insert after existing 20s
arr2 = np.array([1,2,4,9,10,14,26])

#from left to right
x2 = np.searchsorted( arr2 , 5)
print(x2)

# from right to left
y2 = np.searchsorted( arr2 , 15 , side = "right")
print(y2)


# Insert 6 : 1, 2, 4, [6], 9, 10, 14, 26
#                  Index = 3
xx2 = np.searchsorted(arr2 , [5,6,11] , side="right")
print(xx2)
```

#***Arithmetic Functions***

###1. shuffle

Shuffle changes the order of elements randomly.

```python
var1 = np.array([1,2,3,4,5])
np.random.shuffle(var1)
print(var1)
```

###2. unique

returns only the distinct values, removing duplicates.

```python
var2 = np.array([10, 20, 10, 30, 20, 40])

print(np.unique(var2))

# For index
ind = np.unique(var2 , return_index=True)
print(ind)

#For count
count = np.unique(var2 , return_counts=True)
print(count)
```

###3. Resize

np.resize() changes the shape and size of an array.

```python
# np.resize(array, new_shape)
var3 = np.array([1,2,3,4,5,6])

y1 = np.resize(var3, (2,3))
print(y1)
```

###4. Flatten

flatten() converts a multidimensional array into a 1D array.

flatten() creates a copy. Changing the flattened array does not change the original.

```python
var4 = np.array([[1, 2, 3],
                [4, 5, 6]])

new_arr = var4.flatten()
new_arr2 = var4.flatten(order = "F")
print(new_arr)
print(new_arr2)
```

###5. Ravel

ravel() also converts a multidimensional array into a 1D array.

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

new_arr = arr.ravel()

print(new_arr)
```

###Difference between flatten() and ravel()?

Both produce a 1D array, but flatten() always creates a copy, whereas ravel() returns a view when possible. If a view is not possible, ravel() may create a copy.

#Insert


```python
#np.insert(array, index, value, axis=None)

var5 = np.array([1,2,3,4])
print(var5)
print()

v = np.insert(var5, 2, 40)
print(v)

print()
#insert value at 2 positions
v2 = np.insert(var5, (2,4) , 40)
print(v2)

print()
#Float value can't be inserted
v3 = np.insert(var5 , 2 , 6.8)
print(v3)
```

```python
# for 2D

var6 = np.array([[1,2,3], [1,2,3]])

v4 = np.insert(var6,2, 6, axis = 0)
print(v4)

print()
v5 = np.insert(var6,2, 6, axis = 1)
print(v5)

print()
v6 = np.insert(var6,2, [22,23], axis = 1)
print(v6)
```
